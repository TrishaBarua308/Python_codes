```python
print(I)
```

```python
I2 = io.imread("cat.jpeg")
print(I2.shape)
```

```python
plt.rcParams["figure.figsize"] = (3, 3)
io.imshow(I2)
plt.axis("off")
```

```python
plt.rcParams["figure.figsize"] = (15, 5)
plt.subplot(1, 3, 1)
io.imshow(I2[:, :, 0]) # all rows, all columns, 0th channel = Red
plt.title("Red")
plt.axis("off")

plt.subplot(1, 3, 2)
io.imshow(I2[:, :, 1]) # all rows, all columns, 1th channel = Green
plt.title("Green")
plt.axis("off")

plt.subplot(1, 3, 3)
io.imshow(I2[:, :, 2]) # all rows, all columns, 2nd channel = Blue
plt.title("Blue")
plt.axis("off")
```

**Image Cropping**

```python
I_cropped = I2[:, :512, :]
plt.rcParams["figure.figsize"] = (2, 2)
io.imshow(I_cropped)
plt.axis("off")
```

```python
I_cropped_2 = I2[:, 512:, :]
plt.rcParams["figure.figsize"] = (2, 2)
io.imshow(I_cropped_2)
plt.axis("off")
```

```python
I_cropped_3 = I2[200:550, 250:750, :]
plt.rcParams["figure.figsize"] = (2, 2)
io.imshow(I_cropped_3)
plt.axis("off")
```

**Mirror Images**

```python
io.imshow(I2[::-1, :, :]) # Mirror w.r.t horizontal axis
plt.axis("off")
```

```python
io.imshow(I2[:, ::-1, :]) # Mirror w.r.t vertical axis
plt.axis("off")
```

**Image Sampling**

```python
io.imshow(I2[::16, ::16, :])
plt.axis("off")
```
